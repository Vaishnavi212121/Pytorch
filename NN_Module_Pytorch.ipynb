```python

import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
#dataset
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
```

```python
df.head()
```

#train-Test Split

df.iloc[:, 1:] → all columns except the first → features (X)
df.iloc[:, 0] → first column → target (y)
test_size=0.2 → 20% test data and 80% training data.

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

#Scaling

Calculates the mean and standard deviation from training data and scales it.
scaled value = (value - mean) / standard deviation

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
y_train
```

#Label Encoding

```python
encoder = LabelEncoder(). #Creates a label encoder.
y_train = encoder.fit_transform(y_train) #Converts labels
y_test = encoder.transform(y_test) #Converts test labels using the same mapping.
```

```python
y_train
```

#Numpy Arrays To Pytorch Tensors

```python
X_train_tensor = torch.from_numpy(X_train.astype(np.float32)) #Converts training features from NumPy array to PyTorch tensor.
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))   #Converts test features to a PyTorch tensor.
y_train_tensor = torch.from_numpy(y_train.astype(np.float32)) #Converts training labels to PyTorch tensor.
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))   #Converts test labels to PyTorch tensor.
```

```python
y_train_tensor.shape #Shows the shape of training labels.
```

#Defining The Model

```python
#create model class
import torch.nn as nn #nn module


class MySimpleNN(nn.Module):

  def __init__(self, num_features):           #constructor ,num_features-featureswill come in nn at input time

    super().__init__()                       #invoke Parent class constructor [child class constructor call parent class constructor through super keyword]
    self.linear = nn.Linear(num_features, 1) #linear layer (how much features comes in i/p ,hpw much features going as output)
    self.sigmoid = nn.Sigmoid()              #Activation Function

  def forward(self, features):               #forward pass
    out = self.linear(features)              #z=wx
    out = self.sigmoid(out)

    return out
```

#Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

```python
# define loss function
loss_function = nn.BCELoss()  #binnary class entrophy class
```

#Training Pipeline

```python
# create model
model = MySimpleNN(X_train_tensor.shape[1])

# define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model(X_train_tensor)

  # loss calculate
  loss = loss_function(y_pred, y_train_tensor.view(-1,1)) #view=reshape

  # clear gradients
  optimizer.zero_grad()

  # backward pass
  loss.backward()

  # parameters update
  optimizer.step()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

#Evaluation

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.5).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
