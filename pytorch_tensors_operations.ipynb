#comparison opern




```python
j = torch.randint(size=(3,3),low=0, high=10)
k = torch.randint(size=(3,3),low=0, high=10)

print(j)
print(k)
```

```python
# greater than
j > k
# less than
j < k
# equal to
j == k
# not equal to
j != k
# greater than equal to
j >= k
# less than equal to
j <= k
```
